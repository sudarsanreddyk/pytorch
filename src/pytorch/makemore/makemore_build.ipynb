One-hot encoding is a preprocessing technique used to convert categorical variables (text or labels like "Red", "Male", "Paris") into a numerical format that machine learning models can process.

Each category is represented as a binary vector: all values are 0 except for one position, which is 1.

```python
import torch.nn.functional as F
xenc= F.one_hot(xs, num_classes=27).float()
xenc
```

```python
# forward pass
logits= (xenc @ w) # log-counts (predicts log counts)
counts= logits.exp() # equivalent N
probs= counts/counts.sum(1, keepdims=True) # probabilities for next character
probs # every row if we sum, it will be 1
# the last 2 lines are called Softmax activation function
```

```python
# --- 2. CREATE THE DATASET ---
xs, ys = [], []
for w in words: 
    chs = ['.'] + list(w) + ['.'] # Add markers: ['.','t','e','s','t','.']
    
    # Create pairs (Input, Target)
    for ch1, ch2 in zip(chs, chs[1:]):
        ix1 = stoi[ch1] # e.g., '.' -> 0
        ix2 = stoi[ch2] # e.g., 't' -> 1
        xs.append(ix1)  # Input vector
        ys.append(ix2)  # Target vector

xs = torch.tensor(xs)
ys = torch.tensor(ys)
num = xs.nelement() 
print('number of examples: ', num)
```

```python
# --- 3. INITIALIZE THE 'NETWORK' ---
g = torch.Generator().manual_seed(2147483647)
# W is a random matrix. Shape (27, 27) because we have ~27 characters (a-z + dot)
W = torch.randn((27, 27), generator=g, requires_grad=True)
```

```python
# --- 4. GRADIENT DESCENT (Training Loop) ---
for k in range(200): # Run 10 epochs
    
    # Forward pass
    # THIS IS THE ONE-HOT ENCODING STEP
    xenc = F.one_hot(xs, num_classes=27).float() 
    
    logits = xenc @ W # Predict log-counts (Matrix multiplication)
    counts = logits.exp() # Exponentiate to get positive counts
    probs = counts / counts.sum(1, keepdims=True) # Normalize to get probabilities
    
    # Calculate Loss (Negative Log Likelihood)
    loss = -probs[torch.arange(num), ys].log().mean()
    print(loss.item())

    # Backward pass
    W.grad = None # Clear old gradients
    loss.backward() # Compute new gradients

    # Update weights (Manual Gradient Descent)
    W.data += -50 * W.grad
```

```python
# finally, sample from the 'neural net' model
g = torch.Generator().manual_seed(2147483647)

for i in range(5):
    out = []
    ix = 0
    while True:

        # ============
        # BEFORE:
        #p = P[ix]
        # NOW:
        # ============
        xenc = F.one_hot(torch.tensor([ix]), num_classes=27).float()
        logits = xenc @ W # predict log-counts
        counts = logits.exp() # counts, equivalent to N
        p = counts / counts.sum(1, keepdims=True) # probabilities for next character
        # ============

        ix = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
        out.append(itos[ix])
        if ix == 0:
            break
    print('.'.join(out))
```
